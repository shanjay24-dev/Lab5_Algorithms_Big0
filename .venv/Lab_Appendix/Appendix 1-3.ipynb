```python
print("--- EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH ---")
original_dataset = [51, 56, 61, 66, 71, 76, 81, 86, 91, 96]
seed_digit = 6 # TUPM-26-1116 Last digit
id_sum = 17    # TUPM-26-1116 sum[cite: 1, 12]

# Rotate dataset
rotated_dataset = original_dataset[seed_digit:] + original_dataset[:seed_digit]
target_incident = (id_sum * 7) % 100  # 19

# Linear Search
comparisons = 0
found_index = -1
for i in range(len(rotated_dataset)):
    comparisons = comparisons + 1
    if rotated_dataset[i] == target_incident:
        found_index = i
        break

linear_comps = 0
for i in range(len(rotated_dataset)):
    linear_comps = linear_comps + 1
    if rotated_dataset[i] == target_incident:
        break

print("Original:", original_dataset)
print("Rotated:", rotated_dataset)
print("Target:", target_incident, "| Index:", found_index, "| Comparisons:", comparisons)
print("Linear Search Comparisons:", linear_comps)
print("Time & Space Complexity: Time: O(n) | Space: O(1)\n")


print("--- EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS ---")
dept_a = [17, 23, 29, 35, 41]
dept_b = [6, 13, 20, 27, 34]

consolidated = []
i = 0
j = 0
merge_comps = 0

while i < len(dept_a) and j < len(dept_b):
    merge_comps = merge_comps + 1
    if dept_a[i] < dept_b[j]:
        consolidated.append(dept_a[i])
        i = i + 1
    else:
        consolidated.append(dept_b[j])
        j = j + 1

while i < len(dept_a):
    consolidated.append(dept_a[i])
    i = i + 1
while j < len(dept_b):
    consolidated.append(dept_b[j])
    j = j + 1

print("Dept A:", dept_a)
print("Dept B:", dept_b)
print("Consolidated:", consolidated)
print("Before Count:", len(dept_a) + len(dept_b), "| After Count:", len(consolidated))
print("Comparison Count:", merge_comps)
print("Time & Space Complexity: Time: O(n + m) | Space: O(n + m)\n")


print("--- EXERCISE 3: E-COMMERCE ORDER PROCESSING ---")
orders = [45, 12, 88, 33, 7, 56]
data = orders.copy()
bubble_comps = 0
swaps = 0
n = len(data)

for x in range(n):
    for y in range(0, n - x - 1):
        bubble_comps = bubble_comps + 1
        if data[y] > data[y + 1]:
            temp = data[y]
            data[y] = data[y + 1]
            data[y + 1] = temp
            swaps = swaps + 1

print("Original Orders:", orders)
print("Sorted Orders:", data)
print("Comparison Counts:", bubble_comps)
print("Movement Counts (Swaps):", swaps)
print("Time & Space Complexity: Time: O(n^2) | Space: O(1)")
```
